```python
from scipy.special import iv, kv
import matplotlib.pyplot as plt
import numpy as np

xrange = np.linspace(0,1,100)


plt.figure()
plt.plot(xrange, iv(0,xrange), label=r'$I_0 (x)$')
plt.plot(xrange, kv(0,xrange), label=r'$I_0 (x)$')
plt.plot(xrange, iv(1,xrange), label=r'$I_1 (x)$')
plt.plot(xrange, kv(1,xrange), label=r'$I_1 (x)$')

plt.xlabel(r'$x$')
plt.ylabel(r'$I_\nu (x)$  ,  $K_\nu (x)$')
plt.legend()
plt.show()
```

```python
Rg = 30.
Rd = 0.1*Rg
G = 4.5e-6
Sigma0 = 0.1

rrange = np.linspace(1e-6,Rg,100)
yrange = rrange/(2*Rd)

v = np.sqrt(4*np.pi*G*Sigma0*Rd*yrange**2*(iv(0,yrange)*kv(0,yrange) - iv(1,yrange)*kv(1,yrange)))
Mr = 2*np.pi*Sigma0*Rd**2*(1-(1+rrange/Rd)*np.exp(-rrange/Rd))
vK = np.sqrt(G*Mr/rrange)

plt.figure()
plt.plot(rrange, v, label=r'Exponential Disk')
plt.plot(rrange, vK, label=r'Keplerian Equivalent')

plt.xlabel(r'$r$')
plt.ylabel(r'$v_c(r)$')
plt.legend()
plt.show()

```
