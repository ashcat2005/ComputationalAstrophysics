## Midpoint Method

Consider the following system of ODEs,

\begin{equation}
\frac{d \bold{y}}{dt} = \bold{f}(x, \bold{y}).
\end{equation}

The midpoint method is based on the centered differences,

\begin{equation}
\bold{y} (x+h) = \bold{y} (x-h) + 2h \bold{f} (x, \bold{y} (x)) + \mathcal{O} (h^2).
\end{equation}

This can be written as the iterative relation

\begin{equation}
\bold{y}_{n} = \bold{y}_{n-2} + 2h\bold{f}_{n-1} 
\end{equation}

It is clear that this algorithm does not self-inizialize, and therefore, we may use single Euler steps to begin. In order to illustrate the process, consider the integration of the ODEs system from $x=x_0$ to $x_0 + H$. Using $N$ points in this domain, each interval will have a lenght $h=\frac{H}{N}$. Then, we proceed to compute the following iterative steps,

\begin{align}
\bold{y}_1 = &\bold{y}_{0} + h\bold{f}_0   \text{     (Forward Euler step)}\\
\bold{y}_2 = &\bold{y}_{0} + 2h\bold{f}_1 \\
\bold{y}_3 = &\bold{y}_{1} + 2h\bold{f}_2 \\
...\\
\bold{y}_n = &\bold{y}_{n-2} + 2h\bold{f}_{n-1}\\
...\\
\bold{y}_N = &\bold{y}_{N-2} + 2h\bold{f}_{N-1}
\end{align}

The final result can be improved by averaging the midpoint result $\bold{y}_N$ and the estimate obtained by a final Euler formula, $\bold{y}^E_N = \bold{y}_{N-1} + h\bold{f}_N$. This gives

\begin{equation}
\bold{Y}_N = \frac{1}{2}\left[ \bold{y}_N + \bold{y}^E_N \right] 
\end{equation}

\begin{equation}
\bold{Y}_N = \frac{1}{2} \left[ \bold{y}_N + \bold{y}_{N-1} + h\bold{f}_N \right] 
\end{equation}


This is the basis of the *Bulirsch-Stoer Method*, which is the technique of choice in problems where high accuracy is required.

## Richardson Extrapolation

Richardson extrapolation is a simple method for boosting the accuracy of certain numerical procedures, including finite difference approximations. 

Suppose that we have an approximate means of computing some quantity $G$, depending on some parameter or step $h$, in the form
\begin{equation}
G = g(h) + \epsilon (h),
\end{equation}
where $g(h)$ deontes the approximation and $\epsilon(h)$ represents the error of the method. Richardson extrapolation can remove this error, provided that it has the form $\epsilon (h) = c h^p$ with $c$ and $p$ two constants.

The extrapolation begins by computing $g(h)$ with some particular value of $h= h_1$,
\begin{equation}
G = g(h_1) + ch_1^p.
\end{equation}

Then, we repeat the calculation with $h= h_2$, 
\begin{equation}
G = g(h_2) + ch_2^p.
\end{equation}

Eliminating $c$ from this equations and solving for $G$, we obtain the **Richardson extrapolation formula**,
\begin{equation}
G = \frac{\left( \frac{h_1}{h_2}\right)^p g(h_2) - g(h_1)}{\left( \frac{h_1}{h_2} \right)^p - 1}.
\end{equation}

A common practice is to chose $h_1=h$ and $h_2 = \frac{h_1}{2}= \frac{h}{2}$, and therefore
\begin{equation}
G = \frac{ 2^p g(h/2) - g(h)}{2^p - 1}.
\end{equation}

---

Using the Richardson Extrapolation scheme, it is possible to improve the value of the final point in the midpoint method. Since the integration algorithm has an order of approximation of $h^2$ (i.e. $p=2$) we use

\begin{equation}
Y^\text{better}_N =  \frac{ 4 g(h/2) - g(h)}{3},
\end{equation}

where $g$ represents the midpoint method described above and it is needed to calculate the final result using the steps $h$ and $h/2$. This correction gives a fourth-order accurate result.

We can write this result as

\begin{equation}
Y^\text{better}_N =  \frac{ 4 Y_N^{(h/2)} - Y_N^{(h)}}{3},
\end{equation}

Another round of integration with $h/4$ followed by Richardson extrapolation get us sixth-order accuracy, and so forth. Rather than halving the interval in successive integrations, it is often used the sequence $h/2$, $h/4$, $h/6$, $h/8$, $h/10$,..., because it has been found to be more  economical computationally.

## Bulirsch–Stoer Algorithm

When used on its own, the  midpoint method has a major shortcoming: the solution
at points between the initial and final values in the integration domain cannot be refined by the Richardson extrapolation (it is usable only at the last point). 

This deficiency is rectified in the Bulirsch–Stoer method, where the fundamental idea is simple:
Apply the midpoint method in a piecewise fashion. That is, advance the solution in stages of length $H$, using the midpoint method with Richardson extrapolation to perform the integration in each stage. The value of $H$ can be quite large, because the precision of the result is determined by the step length h in the midpoint method, not by $H$. However, if $H$ is too large, the midpoint method may not converge. If this happens, try a smaller value of $H$ or consider a larger error tolerance.

The Runge–Kutta method (fixed step or adaptative step) is more robust, having a higher tolerance for non-smooth functions and stiff problems. The Bulirsch–Stoer algorithm is used mainly in problems where high accuracy is of paramount importance.
