### Abstract

In this notebook we will prepare the data for the example of a Generative Adversary Neural Network (GANN).

```python
import matplotlib.pyplot as plt
import numpy as np

import h5py
from PIL import Image
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Read and separate the images and labels from the file
with h5py.File('/content/drive/MyDrive/Colab Notebooks/Neural Networks/09. Convolutional GANNs/Galaxy10SDSS.h5', 'r') as F:
    images = np.array(F['images'])
    labels = np.array(F['ans'])
```

```python
spiral_galaxies = images[labels>6]
spiral_galaxies.shape
```

```python
plt.figure(figsize=(2,3))
plt.imshow(spiral_galaxies[166])
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
gray_galaxies = np.zeros([2546,69,69])
for i in range(2546):
  img = Image.fromarray(spiral_galaxies[i]) 
  img1 = img.convert('L')
  gray_galaxies[i] = np.asarray(img1)

np.save('/content/drive/MyDrive/Colab Notebooks/Neural Networks/09. Convolutional GANNs/gray_galaxies.npy', gray_galaxies)
```

```python
gray_galaxies_reduced = np.zeros([2546,28,28])
for i in range(2546):
  img = Image.fromarray(spiral_galaxies[i]) 
  img1 = img.convert('L')
  img2 = img1.resize((28,28))
  gray_galaxies_reduced[i] = np.asarray(img2)

np.save('/content/drive/MyDrive/Colab Notebooks/Neural Networks/09. Convolutional GANNs/gray_galaxies_reduced.npy', gray_galaxies_reduced)
```
